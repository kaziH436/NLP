```python
# how to extract emails from a spreadsheet

with open('/kaggle/input/datasets/kazi23/students/students.txt') as f:
    text = f.readlines()
text
```

```python
text=''.join(text)
doc = nlp(text)
emails=[]
for token in doc: 
    if token.like_email:
        emails.append(token.text)
emails
```

```python
from spacy.symbols import ORTH
nlp.tokenizer.add_special_case('gimme', [
    {ORTH: 'gim'},
    {ORTH: 'me'}
])

doc=nlp('gime double cheese')
tokens=[token.text for token in doc]
tokens
```
